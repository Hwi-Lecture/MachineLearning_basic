### 결측치 처리

또 에러가 떴죠?

`Input X contains NaN.`은 입력 데이터에 빈 값이 있다는 뜻입니다. 지금 사용하는 로지스틱 회귀 모델에 넣으려면 이 빈 값을 먼저 처리해야 합니다.

결측치는 다음과 같이 처리할 수 있습니다.

- **행 삭제:** 빈 값이 있는 승객을 제외합니다. 그 승객의 다른 정보도 함께 빠집니다.
- **열 제외:** 해당 항목을 분석에서 제외합니다. 앞에서 제외한 Cabin은 빈 값이 많았던 항목입니다.
- **값 채우기:** 승객과 항목을 유지하고, 빈 칸에 대신 사용할 값을 넣습니다.

현재 모델에 넣을 데이터에서 어느 항목이 얼마나 비어 있는지 확인하겠습니다.

```python
missing_summary = pd.DataFrame({
    '결측치 수': X.isna().sum(),
    '결측 비율(%)': (X.isna().mean() * 100).round(1)
})
missing_summary[missing_summary['결측치 수'] > 0]
```

Age는 약 20%가 비어 있습니다. 나이가 없다는 이유로 행을 삭제하면 해당 승객의 성별, 요금 등 다른 정보도 함께 제외됩니다. 이번에는 승객 정보를 유지하면서 Age와 Embarked의 빈 값을 채우겠습니다.

먼저 나이가 기록된 승객들의 **평균과 중앙값**을 확인합니다. 중앙값은 나이를 작은 순서대로 놓았을 때 가운데에 해당하는 값입니다.

```python
X['Age'].agg(['count', 'mean', 'median']).rename({
    'count': '나이가 기록된 승객 수',
    'mean': '평균 나이',
    'median': '중앙값 나이'
}).to_frame('값')
```

Age의 빈 값은 중앙값으로 채우겠습니다.

Embarked는 탑승 항구를 나타내므로, **가장 많이 등장하는 항구인 최빈값**으로 채우겠습니다. 원래 항구 이름으로 인원을 확인해 보겠습니다. 아래 표의 NaN은 항구가 기록되지 않은 승객입니다.

```python
df['Embarked'].value_counts(dropna=False).rename_axis('탑승 항구').to_frame('승객 수')
```

앞에서 항구를 S=1, C=2, Q=3으로 바꿨으므로, 실제로 채울 때는 해당 항구의 숫자를 사용합니다.

처리 전 값을 복사해 두고, 빈 칸에 넣을 기준값을 확인하겠습니다.

```python
before_fill = X[['Age', 'Embarked']].copy()
age_fill = X['Age'].median()
embarked_fill = X['Embarked'].mode()[0]

print(f'Age에 채울 값: {age_fill}세')
print(f'Embarked에 채울 값: {embarked_fill} (S=1, C=2, Q=3)')
```

`fillna()`는 빈 값만 지정한 값으로 채웁니다. 이미 기록되어 있는 값은 그대로 둡니다.

```python
X['Age'] = X['Age'].fillna(age_fill)
X['Embarked'] = X['Embarked'].fillna(embarked_fill)
```

원래 나이가 비어 있던 승객 5명의 처리 전후를 비교해 보겠습니다.

```python
age_missing = before_fill['Age'].isna()
pd.DataFrame({
    '나이: 처리 전': before_fill.loc[age_missing, 'Age'],
    '나이: 처리 후': X.loc[age_missing, 'Age']
}).head()
```

탑승 항구가 비어 있던 승객도 확인하겠습니다. 처리 후 숫자는 앞에서 정한 항구 코드입니다.

```python
embarked_missing = before_fill['Embarked'].isna()
pd.DataFrame({
    '항구: 처리 전': before_fill.loc[embarked_missing, 'Embarked'],
    '항구: 처리 후': X.loc[embarked_missing, 'Embarked']
})
```

마지막으로 결측치가 모두 채워졌는지, 승객 수는 유지되었는지 확인합니다.

```python
print(f'승객 수: 처리 전 {len(before_fill)}명 → 처리 후 {len(X)}명')
pd.DataFrame({
    '처리 전 결측치 수': before_fill.isna().sum(),
    '처리 후 결측치 수': X[['Age', 'Embarked']].isna().sum()
})
```

이번에야 말로 학습을 해볼 수 있을까요?

```python
model = LogisticRegression(random_state=42)
model.fit(X, y)
```

드디어 학습이 되었습니다! 성능을 한번 확인해 볼까요? 정확도를 한번 측정해 보겠습니다.

그래서 학습용 데이터와 평가용 데이터를 나눠서 모델 예측을 수행하게 됩니다.

학습용 데이터는 모델이 학습할 때 보는 데이터, 평가용 데이터는 모델이 학습할 때 보지 못한 데이터.

이렇게 나눠야 모델이 새로운 데이터를 만났을 때도 잘 맞추는지를 확인할 수 있는거죠.

```python
from sklearn.model_selection import train_test_split

df = pd.read_csv("Titanic-Dataset.csv")

y = df['Survived']
X = df.drop(columns=['Survived'])

X = X.drop(columns=['PassengerId', 'Name', 'Ticket', 'Cabin'])
X['Sex'] = X['Sex'].map({'male': 0, 'female': 1})
X['Embarked'] = X['Embarked'].map({'S': 1, 'C': 2, 'Q': 3})

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# 결측치를 처리할 때 사용했던 결측치 처리를 다시 나눠서 수행합니다.
X_train['Age'] = X_train['Age'].fillna(X_train['Age'].median())
X_train['Embarked'] = X_train['Embarked'].fillna(X_train['Embarked'].mode()[0])

X_test['Age'] = X_test['Age'].fillna(X_train['Age'].median())
X_test['Embarked'] = X_test['Embarked'].fillna(X_train['Embarked'].mode()[0])

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

다시, 훈련셋으로 모델을 훈련하고 테스트셋으로 모델 성능을 예측해 보겠습니다.

```python
model = LogisticRegression(random_state=42)
model.fit(X_train, y_train)

model.score(X_test, y_test)
```

처음 본 데이터에 대해서도 80%의 정확도를 달성하네요. 앞의 결과보다 좀 더 신빈성 있는 결과를 얻은 것 같습니다.

한 번 직접 값을 입력해서 해당 승객의 생존 여부 예측 결과를 받아 봅시다.

```python
# 아래 값을 바꿔가면서 실행해 보세요
new_customer = pd.DataFrame([{
    "Pclass": 0,
    "Sex": 0,
    "Age": 0,
    "SibSp": 0,
    "Parch": 0,
    "Fare": 0,
    "Embarked": 0
}])

# 앞에서 선택한 입력 열과 같은 순서로 모델에 전달합니다.
new_prediction = model.predict(new_customer)[0]
print("모델의 예측:", {0: "사망", 1: "생존"}[new_prediction])
```
