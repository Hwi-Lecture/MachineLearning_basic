# 실습 — Diamonds 회귀, 직접 해보기

`3. regression_diamonds.ipynb`에서 했던 과정을 이번엔 직접 채워봅니다. 아래 각 단계에는 무엇을 해야 하는지와, ChatGPT/Gemini 등에 물어볼 때 쓸 키워드·입력/출력 변수명을 적어뒀습니다. 코드는 스스로 작성해도 되고, AI에게 요청해서 채워도 됩니다.

이번엔 한 번의 train/test 분리 대신 **K-fold 교차검증**으로 성능을 측정하고, 그 위에서 두 가지 개선 방법을 적용해 베이스라인과 비교합니다.

## 1. 데이터 불러오기

seaborn에 내장된 diamonds 데이터셋을 불러오세요.

- 프롬프트 키워드: `seaborn`, `load_dataset("diamonds")`
- 출력 변수명: `df`

## 2. 범주형 변수 인코딩

`cut`, `color`, `clarity`는 등급이 있는 순서형 변수입니다. 아래 순서대로 정수로 매핑(순서형 인코딩)하세요.

- `cut`: Fair < Good < Very Good < Premium < Ideal
- `color`: J < I < H < G < F < E < D
- `clarity`: I1 < SI2 < SI1 < VS2 < VS1 < VVS2 < VVS1 < IF

- 프롬프트 키워드: `순서형 인코딩`, `ordinal encoding`, `map`
- 입력 변수명: `df`
- 출력 변수명: `X`(price를 제외한 입력 변수들), `y`(price)

## 3. K-fold 교차검증 준비

데이터를 한 번만 나누는 대신, 5개로 나눠 돌아가며 학습/평가하는 K-fold 방식을 준비합니다.

- 프롬프트 키워드: `KFold`, `n_splits=5`, `shuffle=True`, `random_state=42`
- 출력 변수명: `kf`

## 4. 베이스라인 모델 — K-fold 평가

선형회귀를 `kf`로 교차검증해서, 폴드별 MAE/RMSE와 그 평균을 구하세요. (이후 비교의 기준값이 됩니다.)

- 프롬프트 키워드: `LinearRegression`, `cross_val_score`, `scoring='neg_mean_absolute_error'`, `scoring='neg_root_mean_squared_error'`
- 입력 변수명: `X`, `y`, `kf`
- 출력 변수명: `baseline_mae_scores`, `baseline_rmse_scores`(폴드별 점수), `baseline_mae`, `baseline_rmse`(평균값)

## 5. 직접 해보기 ① — 파생 변수 만들기

`x`, `y`, `z`(가로/세로/깊이)를 곱해서 부피를 나타내는 새 변수를 만들고, `x`, `y`, `z` 대신 이 변수 하나만 넣은 데이터로 같은 K-fold 평가를 다시 해보세요.

- 프롬프트 키워드: `파생 변수`, `x * y * z`, `컬럼 추가 후 기존 컬럼 제거`
- 입력 변수명: `X`
- 출력 변수명: `volume` 컬럼을 추가하고 `x`,`y`,`z`는 뺀 `X_v` / 같은 방식의 `volume_mae_scores`, `volume_rmse_scores`, `volume_mae`, `volume_rmse`
- 비교 기준: 4번에서 구한 `baseline_mae`, `baseline_rmse`

## 6. 직접 해보기 ② — 이상치가 있는 컬럼 확인

`X_v`의 컬럼들 중에 이상치가 있는 컬럼이 있는지 먼저 확인만 해보세요. (아직 처리는 하지 않습니다.)

- 프롬프트 키워드: `describe`, `boxplot`, `IQR`, `이상치 확인`
- 입력 변수명: `X_v`
- 확인할 것: 어떤 컬럼에 이상치가 있어 보이는지, 몇 개 정도인지 (그래프나 통계로 눈으로 확인하는 단계입니다)

## 7. 이상치 처리 방법 정하기

6번에서 찾은 이상치를 어떻게 처리할지 직접 정해서 적용해보세요. 방법은 자유롭게 선택하면 됩니다.

- 예시 방법: 이상치가 있는 행을 통째로 제거 / 일정 범위를 벗어나는 값을 상한·하한값으로 대체(capping) / 이상치가 심한 컬럼만 로그 변환
- 프롬프트 키워드: (선택한 방법에 따라) `이상치 제거`, `capping`, `clip`, `IQR 기준 필터링`
- 입력 변수명: `X_v`, `y`
- 출력 변수명: `X_clean`, `y_clean` (행을 제거하지 않고 값만 바꾸는 경우 `y_clean`은 `y`와 동일해도 됩니다)

## 8. 처리된 데이터로 학습하고 성능 확인

`X_clean`, `y_clean`으로 4번과 같은 방식(K-fold)의 평가를 수행하고, 베이스라인과 비교해보세요.

- 프롬프트 키워드: `LinearRegression`, `cross_val_score`, `KFold`
- 입력 변수명: `X_clean`, `y_clean`, `kf`
- 출력 변수명: `clean_mae_scores`, `clean_rmse_scores`, `clean_mae`, `clean_rmse`
- 비교 기준: 4번에서 구한 `baseline_mae`, `baseline_rmse`

## 9. 정리

`baseline_mae`, `volume_mae`, `clean_mae`(그리고 각각의 `rmse`)를 표나 출력문으로 나란히 정리해서, 두 시도가 베이스라인 대비 어떤 효과가 있었는지 비교해보세요.
